**Hinweis:** Der RSS als auch MSE sind keine sinnvollen Maße, um zwei unterschiedliche Datensätze hinsichtlich der Performance zu vergleichen. Der R²-Score definiert als:

$$R^2 = 1 - \frac{RSS}{\text{nicht-normierte Varianz in y}}$$

kann hingegen auch zum Vergleich zwischen Datensätzen verwendet werden. In den meisten Regressionen wird als Standard-Metrik der (adjustierte) R²-Score verwendet. Typischerweise gilt $R^2 \in [0, 1]$. Im Fall von 1 beschreibt das Modell den Datensatz sehr gut. Im Fall von $R^2 = 0$ beschreibt das Modell den Datensatz schlecht.
